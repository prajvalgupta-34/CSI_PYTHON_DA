```python
import numpy as np
```

```python
my_list=[1,2,3]
print(my_list)
print(type(my_list))
my_array=np.array(my_list)
print(my_array)
print(type(my_array))
my_nestedlist=[[1,2,3],[4,5,6],[7,8,9]]
my_matrix=np.array(my_nestedlist)
print(my_matrix)
arr=np.arange(0,11)
print(arr)
```
