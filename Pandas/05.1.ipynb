```python
import pandas as pd

# Creating a 3-level MultiIndex
arrays = [
    ['2023', '2023', '2023', '2023', '2024', '2024', '2024', '2024'],
    ['NY', 'NY', 'LA', 'LA', 'NY', 'NY', 'LA', 'LA'],
    ['Laptops', 'Phones', 'Laptops', 'Phones', 'Laptops', 'Phones', 'Laptops', 'Phones']
]
index = pd.MultiIndex.from_arrays(arrays, names=('Year', 'City', 'Product'))

# Creating the DataFrame
df = pd.DataFrame({
    'Sales': [100, 150, 200, 50, 120, 160, 210, 60],
    'Profit': [20, 30, 40, 10, 25, 35, 45, 12]
}, index=index)

print(df)
```

```python
"""You want to analyze only the sales of Phones in NY, across all years.

Because "Year" is the outermost level, you have to bypass it to filter by "City" and "Product"."""
```

```python
# Using loc requires slicing through the outer levels
idx = pd.IndexSlice
loc_result = df.loc[idx[:, 'NY', 'Phones'], :]
```

```python
# Using xs to target specific levels directly
xs_result = df.xs(key=('NY', 'Phones'), level=('City', 'Product'))
print(xs_result)
```

```python
# Using xs and retaining the original MultiIndex structure
complex_xs = df.xs(
    key=('NY', 'Phones'), 
    level=('City', 'Product'), 
    drop_level=False  # Keeps the queried levels in the index
)
print(complex_xs)
```
