```python
"""DDL
1)CREATE -TABLE/DATABASE  (name of table or database, column names, data types, constraints)
eg)CREATE TABLE employee (
    id INT PRIMARY KEY,
    name VARCHAR(50) NOT NULL,
    age INT,
    salary DECIMAL(10, 2)
);
2)ALTER  TABLE (name of table, column name, data type, constraints)-(ADD,REMOVE,RENAME,MODIFY,UPDATE)
eg)ALTER TABLE employee ADD COLUMN department VARCHAR(50);
3)DROP to delete table or database (name of table or database)
eg)DROP TABLE employee;
4)RENAME  to rename object in database -obj_type obj_name TO new_obj_name; 
eg)RENAME TABLE employee TO staff;
5)TRUNCATE-DELETE ALL RECORDS FROM TABLE (name of table)
eg)TRUNCATE TABLE employee;"""
"""DML
1)INSERT INTO (name of table) VALUES (value1, value2, value3,...)
eg)INSERT INTO employee (id, name, age, salary) VALUES (1, 'John', 30, 50000);
2)UPDATE (name of table) SET column1=value1, column2=value2,...
eg)UPDATE employee SET salary=60000 WHERE id=1;
3)DELETE FROM (name of table) WHERE condition;
eg)DELETE FROM employee WHERE age < 25;
4)SELECT column1, column2,... FROM (name of table) WHERE condition;
eg)SELECT * FROM employee WHERE age > 30;

"""
```

```python
import mysql.connector

# 1. Establish a fresh connection
conn = mysql.connector.connect(
    host="localhost",
    user="root",          # Check this matches your local setup
    password="GabruJawan@1020"   # Check this matches your local setup
)
cursor = conn.cursor()

# 2. Guarantee the database exists and is selected
cursor.execute("CREATE DATABASE IF NOT EXISTS sig_demo;")
cursor.execute("USE sig_demo;")

# 3. DDL: Creating a simple table
create_table_query = """
CREATE TABLE IF NOT EXISTS students (
    id INT AUTO_INCREMENT PRIMARY KEY,
    name VARCHAR(50),
    score INT
);
"""

# 4. Execute the query
cursor.execute(create_table_query)
print("DDL Executed: 'students' table is created successfully!")
```

```python
# Insert records one by one
cursor.execute("INSERT INTO students (name, score) VALUES ('Aarav', 85);")
cursor.execute("INSERT INTO students (name, score) VALUES ('Priya', 92);")

# Save the changes permanently to the database
conn.commit()
print("DML Executed: Students added to the table!")
```

```python
# Grab everything from the table
cursor.execute("SELECT * FROM students;")

# Fetch all the results and store them in a variable
records = cursor.fetchall()

print("Current Students in Database:")
for row in records:
    # row[0] is ID, row[1] is name, row[2] is score
    print(f"ID: {row[0]} | Name: {row[1]} | Score: {row[2]}")
```

```python
# Update Aarav's score
cursor.execute("""
    UPDATE students 
    SET score = 95 
    WHERE name = 'Aarav';
""")
conn.commit()
print("Updated Aarav's score.")

# Delete Priya's record
cursor.execute("""
    DELETE FROM students 
    WHERE name = 'Priya';
""")
conn.commit()
print("Deleted Priya's record.")

# Let's see what the table looks like now
cursor.execute("SELECT * FROM students;")
print("\nTable after UPDATE and DELETE:")
for row in cursor.fetchall():
    print(row)
```

```python
# Close the cursor and connection
cursor.close()
conn.close()
print("Database connection closed safely.")
```
