```python
#filter letters starting with a and ending with y
#ay
valid_values=[]
options=["any","albany","apple","world","hello"," "]
for x in options :
    if len(x)<=1:
        continue
    elif x[0]!="a" :
        continue
    elif x[-1]!="y":
        continue
    valid_values.append(x)
print(valid_values)
```

```python
options=["any","albany","apple","world","hello"," "]
valid_values=[alpha for alpha in options 
 if len(alpha)>=1
 if  alpha[0] == "a"
 if alpha[-1]=="y"]
print(valid_values)
```

```python
matrix=[[1,2,3],[4,5,6],[7,8,9]]
flattened=[]
for row in matrix:
    for num in row:
        flattened.append(num)
print(flattened)
```

```python
flattened=[num for row in matrix for num in row]
print(flattened)
```

```python
import pprint
printer=pprint.PrettyPrinter()
lst=[]
for a in range(5):
    for b in range(5):
        for c in range(5):
            lst.append((a,b,c))
printer.pprint(lst)
```

```python
lst=[[[ a for a in range(5)] for b in range(5)] for c in range(5)]
printer.pprint(lst)
```
