```python
list1=("apple","banana","grapes")
new_dict=dict.fromkeys(list1,[2])
new_dict["apple"].append(1)
print(new_dict)
```

```python
keys = ['p', 'q', 'r']

# Create a NEW empty list for every single key
new_dict = {key: [] for key in keys}

# Now, appending only affects 'p'
new_dict['p'].append(1)

print(new_dict)
```

# Sets #
**add**

**union**

**intersection**

**discard**

**remove**

**difference**

```python
set1={"apple","orange","banana"}
set1.add("grapes")
print(set1)
```

```python
set2={1,2,3,4,5}
set3={4,5,6,7,8}
```

```python
print(set2.union(set3))
```

```python
print(set2.intersection(set3))
```

```python
print(set2.difference(set3))
```

```python
set2.remove(1)
print(set2)
```

```python
set2.remove(0)
print(set2)
```

```python
set2.discard(0)
print(set2)
```
