```python
def sum_of_natural_numbers(n):
     total=0
     for i in range(1,n+1):
          total=total+i
     return total
```

```python
sum_of_natural_numbers(10)
```

```python
prices={"box_of_ceral":55,"box_of_mangos":100,"box_of_oranges":40}
quantity={"box_of_ceral":2,"box_of_mangos":3,"box_of_oranges":0}
```

```python
money_spent=0
for i in prices:
    money_spent=money_spent+prices[i]*quantity[i]
print(money_spent)
```

```python
def sum_of_digits(n):
    sum=0
    while n>0:
        digit=n%10
        sum=sum+digit
        n=n//10
    return sum
```

```python
sum_of_digits(457)
```

```python
def star(n):
    for i in range(1,n+1):
        print("*"*i)
    for i in range (n-1,0,-1):
        print("*"*i)
```

```python
star(7)
```

```python
def currency_converter():
    print("Welcome to the Global Currency Converter!")
    
    # The infinite loop keeps the menu running
    while True:
        print("\n--- Conversion Menu ---")
        print("1. Convert INR to US Dollars (USD)")
        print("2. Convert INR to UAE Dirham (AED)")
        print("3. Convert INR to Japanese Yen (JPY)")
        print("4. Exit Program")
        
        choice = input("Enter your choice (1-4): ")
        if choice == '4':
            print("Exiting converter. Have a great day!")
            break 
        if choice in ['1', '2', '3']:
            amount_inr = float(input("Enter the amount in Rupees (₹): "))
            
            if choice == '1':
                usd = amount_inr / 83.0
                # The :.2f limits the output to 2 decimal places
                print(f"₹{amount_inr} is approximately ${usd:.2f} USD")
                
            elif choice == '2':
                aed = amount_inr / 22.6
                print(f"₹{amount_inr} is approximately {aed:.2f} AED")
                
            elif choice == '3':
                jpy = amount_inr / 0.55
                print(f"₹{amount_inr} is approximately ¥{jpy:.2f} JPY")
                
        else:
            print("Invalid choice! Please select a number from 1 to 4.")
            
```

```python
num=31
num1=num//2
is_prime=True

for i in range(2,num1+1):
    if num%i==0:
        is_prime= False
        break
if is_prime :
        print(f"{num} is a prime number ")
```

```python

result_list = []

for num in range(2000, 3001):
    # If it's a multiple of 5, skip the rest of the loop and move to the next number
    if num % 5 == 0:
        continue 
    
    # If the code reaches here, we already know it's NOT a multiple of 5
    if num % 7 == 0:
        result_list.append(str(num))

print(", ".join(result_list))

```
