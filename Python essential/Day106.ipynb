## How to Import Jupyter Notebooks as Modules in Python 3.11.9

You have **TWO METHODS** to import code from Day05.ipynb into Day06.ipynb:

### **METHOD 1: Using `%run` (RECOMMENDED for Jupyter notebooks)**
```python
%run Day05.ipynb
```
- **Simplest approach** - executes all cells from Day05.ipynb
- All functions and variables become directly available in your namespace
- Example: `sum_of_natural_numbers(10)` works directly
- **Best for:** Learning, quick prototyping, working within Jupyter

### **METHOD 2: Using `importnb` (For proper Python module imports)**
```python
import importnb
import Day05

# Use with module prefix
result = Day05.sum_of_natural_numbers(10)
```
- Imports Day05 as a proper Python module
- Better for large projects and code organization
- Requires `importnb` package (already installed to your Python 3.11.9)
- **Note:** Only imports code from executed cells
- **Best for:** Production code, package structure, larger projects

### **Summary for Your Workspace:**
- Use **`%run Day05.ipynb`** in Day06.ipynb for simple imports (quickest method)
- Use **`importnb`** when you want proper module namespacing
- Both methods work in Python 3.11.9 with Jupyter notebooks!
